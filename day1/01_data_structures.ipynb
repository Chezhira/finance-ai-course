```python
# --- LISTS: ordered collections ---
account_codes = ["1001", "1100", "2001", "4001", "5001"]
print(account_codes[0])          # "1001" — zero-indexed
print(account_codes[-1])         # "5001" — last item
print(account_codes[1:3])        # ["1100", "2001"] — slice

# Sorting and filtering
balances = [45200, 128400, 67800, 320000]
print(sorted(balances, reverse=True))  # largest first
over_100k = [b for b in balances if b > 100000]
print(over_100k)   # [128400, 320000]

# --- DICTS: key-value pairs (like a lookup table) ---
account_names = {
    "1001": "Cash and Bank",
    "1100": "Accounts Receivable",
    "2001": "Accounts Payable",
    "4001": "Revenue - Sales",
}
print(account_names["1001"])       # "Cash and Bank"
print(account_names.get("9999", "Not found"))  # safe lookup

# Iterating over a dict
for code, name in account_names.items():
    print(f"  {code}: {name}")

# --- TUPLES: immutable pairs ---
# Good for GL entries (debit/credit must not be mutated)
gl_entry = ("2025-12-31", "1001", 50000, "DR")
date, account, amount, side = gl_entry   # unpacking

# --- SETS: unique values only ---
all_entities = ["EntityA", "EntityB", "EntityA", "EntityA"]
unique_entities = set(all_entities)
print(unique_entities)  # {"EntityA", "EntityB"}
```
