```python
def calculate_variance(actual: float, budget: float) -> dict:
    """Calculate variance between actual and budget amounts."""
    variance = actual - budget
    pct = (variance / budget * 100) if budget != 0 else 0
    flag = "OVER" if variance > 0 else "UNDER"
    return {
        "actual": actual,
        "budget": budget,
        "variance": variance,
        "variance_pct": round(pct, 2),
        "flag": flag,
    }

def classify_account(code: str) -> str:
    """Classify account by code prefix (IFRS-style chart of accounts)."""
    prefix = int(str(code)[0])
    mapping = {1: "Asset", 2: "Liability", 3: "Equity",
               4: "Income", 5: "COGS", 6: "OpEx", 7: "Finance"}
    return mapping.get(prefix, "Unknown")

def net_profit(revenue: float, cogs: float, opex: float) -> dict:
    gross = revenue - cogs
    net = gross - opex
    return {
        "revenue": revenue,
        "gross_profit": gross,
        "gross_margin_pct": round(gross / revenue * 100, 1),
        "net_profit": net,
        "net_margin_pct": round(net / revenue * 100, 1),
    }

# --- Run it ---
result = calculate_variance(actual=410000, budget=390000)
print(f"Revenue variance: ${result['variance']:,.0f} ({result['variance_pct']}%) — {result['flag']}")

pnl = net_profit(revenue=462000, cogs=220000, opex=118600)
print(f"Net profit: ${pnl['net_profit']:,.0f} ({pnl['net_margin_pct']}% margin)")
```
